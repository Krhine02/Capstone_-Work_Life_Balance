```python
# Dropping Columns

df = df.drop(columns=['Employee_ID', 'Work_Life_Balance_Rating', 'Productivity_Change', 'Number_of_Virtual_Meetings', 'Social_Isolation_Rating', 'Satisfaction_with_Remote_Work', 'Company_Support_for_Remote_Work', 'Region'])
df.head()
```

```python
df = df.fillna("None")
```

```python
print(df.isna().sum())
```

```python
df = df.rename(columns={
    'Job_Role': 'Job_Tittle',
    'Industry': 'Occupation',
    'Work_Location': 'Work_Enviorment'
})
```

```python
# What age group reports metal health conditions frequently in what work environment?

filtered_df = df[
    (df["Age"] <= 40) &
    (df["Mental_Health_Condition"] .isin(["Depression", "Anxiety", "Burnout"]))
]

print(filtered_df)
print("Number of People:", len(filtered_df))
```

```python
filtered_df = df[
    (df["Age"] <= 40) &
    (df["Mental_Health_Condition"] .isin(["None"]))
]

print(filtered_df)
print("Number of People:", len(filtered_df))
```

```python
filtered_df = df[
    (df["Age"] >= 40) &
    (df["Mental_Health_Condition"] .isin(["Depression", "Anxiety", "Burnout"]))
]

print(filtered_df)
print("Number of People:", len(filtered_df))
```

```python
filtered_df = df[
    (df["Age"] >= 40) &
    (df["Mental_Health_Condition"] .isin(["None"]))
]

print(filtered_df)
print("Number of People:", len(filtered_df))
```

```python
# What occupations have the most access to mental health resources?/ Value_counts
occupation_resources = df[df["Access_to_Mental_Health_Resources"] == "Yes"]

print(occupation_resources["Occupation"].value_counts())
```

```python
# What Job titles reports Poor work sleep quality?/ Groupby

Job_Title = (
    df[df["Sleep_Quality"] == "Poor"]
    .groupby ("Job_Tittle")
    .size()
    .sort_values(ascending=False)
)

print(Job_Title)
```

```python
import matplotlib.pyplot as plt
```

```python
Job_Title.plot(kind="barh")


plt.xlabel("Number of Reports")
plt.ylabel("Job Title")
plt.title("Job Titles Reporting Poor sleep Quality")


plt.savefig("Job_Title.png", bbox_inches="tight")
plt.show()
```

```python
df["Work_Enviorment"].value_counts()
```

```python
age_20_40 = df[(df["Age"] >= 20) & (df["Age"] <= 40)]

age_41_60 = df[(df["Age"] > 41) & (df["Age"] <= 60)]
```

```python
work_20_40 = age_20_40["Work_Enviorment"].value_counts()
work_41_60 = age_41_60["Work_Enviorment"].value_counts()

comparison = pd.DataFrame({
    "Ages 20-40": work_20_40,
    "Ages 41-60": work_41_60
}).fillna(0)

comparison.plot(
    kind="bar",
    color=["blue", "red"]
)
plt.xlabel("Work Environment")
plt.ylabel("Number of People")
plt.title("Work Environment by Age Group")
plt.xticks(rotation=0)
plt.legend()


plt.show()
```

```python
import matplotlib.pyplot as plt
mental_health_work = pd.crosstab(
    df["Work_Enviorment"],
    df["Mental_Health_Condition"]
)

mental_health_work = mental_health_work.drop(columns="None")

ax =mental_health_work.plot(kind="barh", stacked=True)

totals = mental_health_work.sum(axis=1)

for i, total in enumerate(totals):
    ax.text(total + 1, i, str(total), va="center")

plt.xlabel("Number of Reports")
plt.ylabel("Work Environment")
plt.title("Mental Health Conditions by Work Environment")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = df[df["Age"].between(41, 60)]

data = data[data["Work_Enviorment"] != "Hybrid"]

chart = pd.crosstab(
    data["Work_Enviorment"],
    data["Sleep_Quality"]
)


sns.heatmap(chart, annot=True, fmt="d")

plt.title("Sleep Quality by Work Environment (Ages 41-60)")
plt.show()

```
